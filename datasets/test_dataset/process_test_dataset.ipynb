```python
import pandas as pd

localize_df = pd.read_csv("test_dataset_localization.csv")

localize_df = localize_df[["item_index", "source_code", "vuln_lines"]]

for index, row in localize_df.iterrows():
    localize_df.loc[index, "vuln_lines"] = "[" + ", ".join(str(i) for i in list(eval(row["vuln_lines"]))) + "]"

localize_df.to_csv("custom_localize.csv", index=False, encoding='utf-8')
```

```python
import pandas as pd

detect_df = pd.read_csv("test_dataset_detection.csv")

df = pd.DataFrame()
df['item_index'] = [i for i in range(len(detect_df))]
df['source_code'] = detect_df['source_code']
df['vuln_types'] = detect_df['vul_type']

for index, row in df.iterrows():
    df.loc[index, "vuln_types"] = "[" + row["vuln_types"][1:-1] + "]"

df.to_csv("custom_detect.csv", index=False, encoding='utf-8')
```
