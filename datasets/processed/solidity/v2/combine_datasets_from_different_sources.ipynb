```python
import pandas as pd

defi_vuln = pd.read_csv('defi_vuln_labs_dataset_with_vuln_line_numbers.csv')
vuln_verify = pd.read_csv('vulnerable_verified_smart_contracts_dataset_with_line_numbers_with_source_code.csv')
smartbugs = pd.read_csv('smartbugs_curated_dataset_with_vuln_line_numbers_v2.csv')
```

```python
display(vuln_verify.head())
print(len(vuln_verify))
```

```python
display(defi_vuln.head())
print((defi_vuln['done'] == 1).sum())
```

```python
display(smartbugs.head())
print(len(smartbugs))
```

```python
vuln_verify['source'] = "vulnerable_verified_dataset"
for index, row in defi_vuln.iterrows():
    if row['done'] == 1:
        file_path = row['file']
        contract_address = None
        final_line_numbers = row['vuln_lines']
        vuln = None
        type_ = None
        source_code = row['source_code']
        source = "defi_vuln_labs"
        new_row = {'file_path': file_path, 'contract_address': contract_address, 'final_line_numbers': final_line_numbers, 'vuln': vuln, 'type': type_, 'source_code': source_code, 'source': source}
        vuln_verify.loc[len(vuln_verify)] = new_row  # Append the new row at the end

for index, row in smartbugs.iterrows():
    file_path = row['subfolder'] + "\\" + row['filename']
    contract_address = None
    final_line_numbers = row['vuln_lines']
    vuln = row['subfolder']
    type_ = None
    source_code = row['source_code']
    source = "smartbugs_curated"
    new_row = {'file_path': file_path, 'contract_address': contract_address, 'final_line_numbers': final_line_numbers, 'vuln': vuln, 'type': type_, 'source_code': source_code, 'source': source}
    vuln_verify.loc[len(vuln_verify)] = new_row  # Append the new row at the end
```
