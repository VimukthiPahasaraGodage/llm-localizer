# smart contract vulndb

```python
import json

file_path = f"{cwd}\\smart-contract-vulndb\\dataset\\vulns-1.json"

data = []

with open(file_path, encoding="utf-8") as f:
    for line in f:
        try:
            json_obj = json.loads(line)  # Parse JSON from each line
            data.append(json_obj)
        except json.JSONDecodeError as e:
            print(f"Skipping invalid JSON: {e}")

print(f"Loaded {len(data)} JSON objects")
```

```python
import random
from IPython.display import Markdown

random_row = random.randint(0, len(data) - 1)  # Generates a number between 10 and 50 (inclusive)
print("Random row:", random_row)

# Pretty-printing the JSON
pretty_json = json.dumps(data[random_row], indent=4)
print(pretty_json)

# Example Markdown content
md_content = data[random_row]['body']

Markdown(md_content)
```
