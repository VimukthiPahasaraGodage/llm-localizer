```python
import numpy as np
from scipy.stats import wilcoxon

def compare_two_models(model_a_results, model_b_results, alpha=0.05):
    model_a_results = model_a_results/100
    model_b_results = model_b_results/100
    # Perform Wilcoxon signed-rank test (two-sided by default)
    statistic, p_value = wilcoxon(model_a_results, model_b_results)
    
    print(f"Wilcoxon statistic = {statistic:.3f}, p-value = {p_value:.4f}")

    if p_value < 0.05:
        print("Statistically significant — meaning one model performs better consistently across folds")
    else:
        print("No significant difference")
        
    # Determine which model performed better on average
    mean_diff = np.mean(model_a_results - model_b_results)
    print(f"Mean F1 difference (A - B): {mean_diff:.4f}")

    if mean_diff > 0:
        print("First model is better on average")
    else:
        print("Second model is better on average")
```

```python
import pandas as pd
import numpy as np
from scipy.stats import shapiro, friedmanchisquare, ttest_rel
import statsmodels.api as sm
from statsmodels.stats.anova import AnovaRM
import scikit_posthocs as sp
import matplotlib.pyplot as plt
import itertools

# -------- STEP 1: INPUT YOUR DATA --------
# Dictionary of model F1 scores across the same k folds
f1_scores = {
    'Model_A': [0.84, 0.77, 0.78, 0.80, 0.76, 0.79, 0.75, 0.82, 0.78, 0.77],
    'Model_B': [0.73, 0.79, 0.74, 0.75, 0.74, 0.73, 0.72, 0.76, 0.75, 0.68],
    'Model_C': [0.57, 0.73, 0.69, 0.70, 0.68, 0.71, 0.69, 0.72, 0.70, 0.77],
    # Add more models if needed
}

# -------- STEP 2: PREPARE DATA --------
df = pd.DataFrame(f1_scores)
df['Fold'] = range(1, len(df) + 1)
long_df = df.melt(id_vars='Fold', var_name='Model', value_name='F1')

model_names = list(f1_scores.keys())

# -------- STEP 3: NORMALITY CHECK --------
print("🔍 Shapiro-Wilk Normality Test on Pairwise Differences:")
normal = True
for a, b in itertools.combinations(model_names, 2):
    diff = df[a] - df[b]
    stat, p = shapiro(diff)
    print(f"{a} - {b}: p = {p:.4f}")
    if p < 0.05:
        normal = False

# -------- STEP 4: REPEATED MEASURES ANOVA (if normal) --------
if normal:
    print("\n📊 Repeated-Measures ANOVA:")
    aov = AnovaRM(long_df, depvar='F1', subject='Fold', within=['Model'])
    print(aov.fit())
else:
    print("\n⚠️ Normality assumption violated. Skipping ANOVA.")

# -------- STEP 5: FRIEDMAN TEST --------
print("\n📈 Friedman Test (Non-parametric):")
score_arrays = [df[model] for model in model_names]
friedman_stat, friedman_p = friedmanchisquare(*score_arrays)
print(f"Chi-square = {friedman_stat:.4f}, p = {friedman_p:.4f}")

# -------- STEP 6: POST-HOC TEST (Nemenyi) --------
print("\n🔬 Post-hoc Nemenyi Test:")
nemenyi = sp.posthoc_nemenyi_friedman(df[model_names])
print(nemenyi)

# Optional: Critical Difference Plot
sp.sign_plot(nemenyi, alpha=0.05)
plt.title("Nemenyi Critical Difference Diagram")
plt.show()

# -------- STEP 7: DETERMINE BEST MODEL --------
mean_scores = df[model_names].mean()
best_model = mean_scores.idxmax()
print(f"\n🏆 Best mean F1 model: {best_model} ({mean_scores[best_model]:.4f})")

# Check significance vs others
print(f"\n📌 Checking if {best_model} is significantly better than all others:")
top_key = best_model
others = [m for m in model_names if m != top_key]
significant = True
for other in others:
    pval = nemenyi.loc[top_key, other] if top_key in nemenyi.index else nemenyi.loc[other, top_key]
    print(f"{top_key} vs {other}: p = {pval:.4f}")
    if pval >= 0.05:
        significant = False

if significant:
    print(f"\n✅ {top_key} is significantly better than all others — it's the best model.")
else:
    print(f"\n⚠️ {top_key} has the highest mean F1 but is NOT significantly better than all other models.")

```
