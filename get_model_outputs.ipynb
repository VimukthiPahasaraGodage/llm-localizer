```python
import os
import zipfile
import boto3

AWS_ACCESS_KEY_ID = ''
AWS_SECRET_ACCESS_KEY = ''
AWS_REGION = 'us-east-1'
BUCKET_NAME = 'final-fyp-data'

def download_s3_directory(bucket_name, s3_prefix, local_dir):
    # Initialize the S3 client
    s3 = boto3.client(
        's3',
        aws_access_key_id=AWS_ACCESS_KEY_ID,
        aws_secret_access_key=AWS_SECRET_ACCESS_KEY,
        region_name=AWS_REGION
    )

    # Get the list of objects in the S3 bucket at the specified prefix
    paginator = s3.get_paginator('list_objects_v2')
    result = paginator.paginate(Bucket=bucket_name, Prefix=s3_prefix)

    num_files = 0
    for page in result:
        for obj in page.get('Contents', []):
            # Get the file key and generate the local file path
            s3_key = obj['Key']
            if 'logs' in s3_key:
                local_file_path = os.path.join(local_dir, s3_key[len(s3_prefix):]).replace("\\", "/")
                os.makedirs(os.path.dirname(local_file_path), exist_ok=True)
                num_files += 1
                
                # Download the file
                print(local_file_path)
                s3.download_file(bucket_name, s3_key, local_file_path)
                print(f"Downloaded {s3_key} to {local_file_path}")
    return num_files
```
