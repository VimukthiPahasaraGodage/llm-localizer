```python
import os
import csv
from tensorboard.backend.event_processing import event_accumulator

def convert_and_save_tensorboard_event_files_as_csvs(experiment, fold):
    # Path to your TensorBoard logs (folder containing the events.out.tfevents.* files)
    log_dir = f'{os.getcwd()}/{parent_tensorboard_logs_folder}/{experiment}/logs/fold_{fold}/'.replace("\\", "/")
    
    # Load the event file
    ea = event_accumulator.EventAccumulator(log_dir,
        size_guidance={ # see below for explanations
            event_accumulator.SCALARS: 0,
        })
    ea.Reload()
    
    # List all scalar tags (e.g., 'loss', 'accuracy')
    scalar_tags = ea.Tags()['scalars']
    
    # Create output folder if needed
    output_dir = f'{os.getcwd()}/{csv_files_output_folder}/{experiment}/fold_{fold}'
    os.makedirs(output_dir, exist_ok=True)
    
    # For each scalar, save its values to a CSV
    for tag in scalar_tags:
        events = ea.Scalars(tag)
        
        csv_path = os.path.join(output_dir, f"{tag.replace('/', '_')}.csv")
        
        with open(csv_path, mode='w', newline='') as f:
            writer = csv.writer(f)
            writer.writerow(['step', 'wall_time', 'value'])  # Header
            
            for event in events:
                writer.writerow([event.step, event.wall_time, event.value])
    
    print(f"Done saving CSVs for fold: {fold} of experiment: {experiment}")
```

```python
for experiment in experiments:
    for i in range(0, 10):
        convert_and_save_tensorboard_event_files_as_csvs(experiment, i)
```
